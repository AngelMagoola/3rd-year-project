```python
EXTRACTED_DIR = os.path.join(PROJECT_DIR, "dataset", "extracted")

RUST_EXTRACT_DIR = os.path.join(EXTRACTED_DIR, "rust")
HEALTHY_EXTRACT_DIR = os.path.join(EXTRACTED_DIR, "healthy")

os.makedirs(RUST_EXTRACT_DIR, exist_ok=True)
os.makedirs(HEALTHY_EXTRACT_DIR, exist_ok=True)

with zipfile.ZipFile(rust_zip, "r") as z:
    z.extractall(RUST_EXTRACT_DIR)

with zipfile.ZipFile(healthy_zip, "r") as z:
    z.extractall(HEALTHY_EXTRACT_DIR)

print("Extraction completed.")
print("Rust folder:", os.listdir(RUST_EXTRACT_DIR))
print("Healthy folder:", os.listdir(HEALTHY_EXTRACT_DIR))
```

```python
IMAGE_EXTENSIONS = (".jpg", ".jpeg", ".png", ".bmp", ".webp")

def count_images(folder):
    return sum(
        1 for root, dirs, files in os.walk(folder)
        for file in files
        if file.lower().endswith(IMAGE_EXTENSIONS)
    )

rust_count = count_images(RUST_EXTRACT_DIR)
healthy_count = count_images(HEALTHY_EXTRACT_DIR)

print("Rust images:", rust_count)
print("Healthy images:", healthy_count)
print("Total images:", rust_count + healthy_count)
```

```python
from PIL import Image

def check_images(folder):
    valid = 0
    corrupted = []

    for root, dirs, files in os.walk(folder):
        for file in files:
            if file.lower().endswith(IMAGE_EXTENSIONS):
                path = os.path.join(root, file)

                try:
                    with Image.open(path) as img:
                        img.verify()
                    valid += 1
                except Exception:
                    corrupted.append(path)

    return valid, corrupted

rust_valid, rust_corrupted = check_images(RUST_EXTRACT_DIR)
healthy_valid, healthy_corrupted = check_images(HEALTHY_EXTRACT_DIR)

print("RUST")
print("Valid:", rust_valid)
print("Corrupted:", len(rust_corrupted))

print("\nHEALTHY")
print("Valid:", healthy_valid)
print("Corrupted:", len(healthy_corrupted))
```

```python
print("Corrupted healthy image:")
for path in healthy_corrupted:
    print(path)
```

```python
from collections import Counter

def inspect_image_metadata(folder):
    dimensions = Counter()
    formats = Counter()
    total = 0

    for root, dirs, files in os.walk(folder):
        for file in files:
            if file.lower().endswith(IMAGE_EXTENSIONS):
                path = os.path.join(root, file)

                try:
                    with Image.open(path) as img:
                        dimensions[img.size] += 1
                        formats[img.format] += 1
                        total += 1
                except Exception:
                    continue

    return total, dimensions, formats

rust_total, rust_dimensions, rust_formats = inspect_image_metadata(RUST_EXTRACT_DIR)
healthy_total, healthy_dimensions, healthy_formats = inspect_image_metadata(HEALTHY_EXTRACT_DIR)

print("RUST")
print("Images:", rust_total)
print("Formats:", rust_formats)
print("Most common dimensions:", rust_dimensions.most_common(10))

print("\nHEALTHY")
print("Images:", healthy_total)
print("Formats:", healthy_formats)
print("Most common dimensions:", healthy_dimensions.most_common(10))
```

```python
import hashlib
import pandas as pd

def calculate_sha256(path):
    sha = hashlib.sha256()

    with open(path, "rb") as f:
        for chunk in iter(lambda: f.read(8192), b""):
            sha.update(chunk)

    return sha.hexdigest()

records = []

for label, folder in [
    ("rust", RUST_EXTRACT_DIR),
    ("healthy", HEALTHY_EXTRACT_DIR)
]:
    for root, dirs, files in os.walk(folder):
        for file in files:
            if file.lower().endswith(IMAGE_EXTENSIONS):
                path = os.path.join(root, file)

                records.append({
                    "filename": file,
                    "class": label,
                    "image_path": path,
                    "sha256": calculate_sha256(path)
                })

metadata_df = pd.DataFrame(records)

print("Images hashed:", len(metadata_df))
print("Unique SHA-256 hashes:", metadata_df["sha256"].nunique())
print("Duplicate image instances:", len(metadata_df) - metadata_df["sha256"].nunique())
```

```python
# Count images and unique hashes in each class
print("IMAGE COUNTS BY CLASS:")
print(metadata_df.groupby("class").size())

print("\nUNIQUE IMAGES BY CLASS:")
print(metadata_df.groupby("class")["sha256"].nunique())

# Identify duplicate groups
duplicate_groups = (
    metadata_df.groupby("sha256")
    .filter(lambda group: len(group) > 1)
    .sort_values("sha256")
)

print("\nDUPLICATE INSTANCES BY CLASS:")
print(duplicate_groups.groupby("class").size())

# Identify hashes appearing in both classes
cross_class_hashes = (
    metadata_df.groupby("sha256")["class"]
    .nunique()
)

cross_class_hashes = cross_class_hashes[cross_class_hashes > 1]

print("\nDUPLICATE GROUPS SHARED ACROSS CLASSES:")
print(len(cross_class_hashes))

# Recheck the corrupted file
print("\nCORRUPTED FILE STILL EXISTS:", os.path.exists(corrupted_path))
print("CURRENT IMAGE COUNT:", len(metadata_df))
```
